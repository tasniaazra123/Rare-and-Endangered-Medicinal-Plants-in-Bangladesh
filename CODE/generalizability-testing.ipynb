```python
# =========================
# Grad-CAM Implementation for DenseNet121
# =========================

import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers, models
from tensorflow.keras.applications import DenseNet121
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.metrics import classification_report, confusion_matrix
from sklearn.model_selection import train_test_split
from sklearn.utils.class_weight import compute_class_weight
import os
import time
import cv2
from PIL import Image
import warnings
warnings.filterwarnings('ignore')

# Set random seed for reproducibility
RANDOM_SEED = 42
tf.random.set_seed(RANDOM_SEED)
np.random.seed(RANDOM_SEED)

# =========================
# GPU Configuration
# =========================
gpus = tf.config.list_physical_devices('GPU')
if gpus:
    try:
        for gpu in gpus:
            tf.config.experimental.set_memory_growth(gpu, True)
        print(f"GPU is available: {gpus[0]}")
    except RuntimeError as e:
        print("Error setting memory growth:", e)
else:
    print("No GPU detected.")

# =========================
# Dataset Configuration
# =========================
dataset_path = "/kaggle/input/rare-and-endangered-medicinal-plants-in-bangladesh/Raw-Images"
IMG_SIZE = (224, 224)
BATCH_SIZE = 32
EPOCHS = 100
VALIDATION_SPLIT = 0.2
TEST_SPLIT = 0.15

plant_mapping = {
    'type_1': 'Kalochitra',
    'type_2': 'Anantamu', 
    'type_3': 'Apang',
    'type_4': 'Syspan',
    'type_5': 'Appishikha',
    'type_6': 'Kalamegh',
    'type_7': 'Kalothutura',
    'type_8': 'Gshnru',
    'type_9': 'Chapalish',
    'type_10': 'Beta',
    'type_11': 'Punarnava',
    'type_12': 'Bamonhati',
    'type_13': 'Basok',
    'type_14': 'Ramtulsi',
    'type_15': 'Shotomuli',
    'type_16': 'Sarpagandha'
}

# =========================
# Data Loading Function
# =========================
def load_dataset():
    """Load and preprocess the dataset"""
    
    print("Loading dataset...")
    
    images = []
    labels = []
    class_names = []
    
    # Verify dataset path
    if not os.path.exists(dataset_path):
        print(f"Error: Dataset path '{dataset_path}' does not exist!")
        return None, None, None
    
    # Load images
    total_images = 0
    for type_folder, plant_name in plant_mapping.items():
        folder_path = os.path.join(dataset_path, type_folder)
        
        if os.path.exists(folder_path):
            image_files = [f for f in os.listdir(folder_path) 
                          if f.lower().endswith(('.jpg', '.jpeg', '.png'))]
            
            print(f"Loading {len(image_files)} images from {plant_name}...")
            
            for img_file in image_files:
                img_path = os.path.join(folder_path, img_file)
                try:
                    img = Image.open(img_path).convert('RGB')
                    img = img.resize(IMG_SIZE)
                    img_array = np.array(img, dtype=np.float32)
                    
                    images.append(img_array)
                    labels.append(list(plant_mapping.keys()).index(type_folder))
                    
                except Exception as e:
                    print(f"Error loading {img_path}: {e}")
                    continue
            
            class_names.append(plant_name)
            total_images += len(image_files)
    
    if not images:
        print("No images were loaded.")
        return None, None, None
    
    # Convert to numpy arrays
    images = np.array(images, dtype=np.float32)
    labels = np.array(labels, dtype=np.int32)
    
    print(f"\nDataset Summary:")
    print(f"Total images: {len(images)}")
    print(f"Number of classes: {len(class_names)}")
    print(f"Image shape: {images[0].shape}")
    
    return images, labels, class_names

# Load the dataset
images, labels, class_names = load_dataset()

# =========================
# Train/Validation/Test Split (SAME as your original code)
# =========================
if images is not None:
    # First split: separate test set
    X_temp, X_test, y_temp, y_test = train_test_split(
        images, labels,
        test_size=TEST_SPLIT,
        random_state=RANDOM_SEED,
        stratify=labels
    )

    # Second split: separate validation set from remaining data
    val_split_ratio = VALIDATION_SPLIT / (1 - TEST_SPLIT)
    X_train, X_val, y_train, y_val = train_test_split(
        X_temp, y_temp,
        test_size=val_split_ratio,
        random_state=RANDOM_SEED,
        stratify=y_temp
    )

    print("\nData Split Summary:")
    print(f"Training set: {X_train.shape[0]} images")
    print(f"Validation set: {X_val.shape[0]} images")
    print(f"Test set: {X_test.shape[0]} images")
    
    # Compute class weights (SAME as your original code)
    print("\nComputing class weights for imbalance handling...")
    unique_classes = np.unique(y_train)
    class_weights_array = compute_class_weight(
        class_weight='balanced',
        classes=unique_classes,
        y=y_train
    )
    class_weights = dict(zip(unique_classes, class_weights_array))
    
    print("Class weights:")
    for cls, weight in class_weights.items():
        class_name = class_names[cls] if cls < len(class_names) else f"Class_{cls}"
        print(f"  {class_name}: {weight:.4f}")
    
    # One-hot encode labels
    num_classes = len(class_names)
    y_train_cat = tf.keras.utils.to_categorical(y_train, num_classes)
    y_val_cat = tf.keras.utils.to_categorical(y_val, num_classes)
    y_test_cat = tf.keras.utils.to_categorical(y_test, num_classes)
    
    # Create TensorFlow datasets
    def create_dataset(X, y, batch_size=32, shuffle=False):
        dataset = tf.data.Dataset.from_tensor_slices((X, y))
        if shuffle:
            dataset = dataset.shuffle(buffer_size=1000, seed=RANDOM_SEED)
        dataset = dataset.batch(batch_size)
        dataset = dataset.prefetch(tf.data.AUTOTUNE)
        return dataset
    
    train_dataset = create_dataset(X_train, y_train_cat, BATCH_SIZE, shuffle=True)
    val_dataset = create_dataset(X_val, y_val_cat, BATCH_SIZE)
    test_dataset = create_dataset(X_test, y_test_cat, BATCH_SIZE)

# =========================
# DenseNet121 Model with Functional API for Grad-CAM access
# =========================
def create_densenet121_model_functional():
    """Create DenseNet121 model using Functional API for Grad-CAM access"""
    
    print("Building DenseNet121 model with Functional API...")
    
    # Input layer
    inputs = keras.Input(shape=(IMG_SIZE[0], IMG_SIZE[1], 3))
    
    # Normalization layers (SAME as your original)
    x = layers.Lambda(lambda x: (x / 255.0))(inputs)
    x = layers.Lambda(lambda x: (x - [0.485, 0.456, 0.406]) / [0.229, 0.224, 0.225])(x)
    
    # Base model
    base_model = DenseNet121(
        weights='imagenet', 
        include_top=False, 
        input_tensor=x
    )
    base_model.trainable = True
    
    # Get the output of the base model (last conv layer for Grad-CAM)
    x = base_model.output  # This is the last convolutional layer output
    
    # Store for Grad-CAM access
    last_conv_layer_output = x
    
    # Classification head (SAME as your original)
    x = layers.GlobalAveragePooling2D(name="gap")(x)
    x = layers.Dropout(0.3, name="dropout_1")(x)
    x = layers.Dense(512, activation='relu', name="dense_512")(x)
    x = layers.BatchNormalization(name="batch_norm_1")(x)
    x = layers.Dropout(0.5, name="dropout_2")(x)
    outputs = layers.Dense(num_classes, activation='softmax', name="output_layer")(x)
    
    # Create the main model for training
    model = keras.Model(inputs=inputs, outputs=outputs, name="densenet121_model")
    
    # Create a model for Grad-CAM that returns both last conv layer and predictions
    grad_cam_model = keras.Model(
        inputs=model.input,
        outputs=[last_conv_layer_output, model.output],
        name="grad_cam_model"
    )
    
    # Compile model 
    model.compile(
        optimizer=keras.optimizers.Adam(learning_rate=0.001),
        loss='categorical_crossentropy',
        metrics=['accuracy']
    )
    
    print("DenseNet121 model built successfully with Grad-CAM support!")
    
    return model, grad_cam_model

# Create the models
model, grad_cam_model = create_densenet121_model_functional()
model.summary()

# =========================
# Training Function (SAME as your original)
# =========================
def train_model(model, train_data, val_data, epochs=100):
    """Train the model with same callbacks as your original"""
    
    callbacks = [
        keras.callbacks.EarlyStopping(
            monitor='val_accuracy',
            patience=10,
            restore_best_weights=True,
            verbose=1
        ),
        keras.callbacks.ReduceLROnPlateau(
            monitor='val_loss',
            factor=0.5,
            patience=5,
            min_lr=1e-6,
            verbose=1
        ),
        keras.callbacks.ModelCheckpoint(
            'best_densenet121_gradcam_model.h5',
            monitor='val_accuracy',
            save_best_only=True,
            verbose=1
        )
    ]
    
    print("Starting training...")
    start_time = time.time()
    
    history = model.fit(
        train_data,
        epochs=epochs,
        validation_data=val_data,
        callbacks=callbacks,
        class_weight=class_weights,  # Using class weights as in your original
        verbose=1
    )
    
    training_time = time.time() - start_time
    print(f"Training completed in {training_time:.2f} seconds")
    
    return history, training_time

# Train the model
print("\n" + "="*60)
print("TRAINING DenseNet121 (Same parameters as original)")
print("="*60)
history, training_time = train_model(model, train_dataset, val_dataset, EPOCHS)

# =========================
# Simplified GRAD-CAM Implementation
# =========================
def make_gradcam_heatmap(img_array, model, last_conv_layer_name, pred_index=None):
    """
    Generate Grad-CAM heatmap for a single image
    
    Args:
        img_array: Input image array (preprocessed)
        model: Keras model
        last_conv_layer_name: Name of the last convolutional layer
        pred_index: Class index to generate heatmap for (None = predicted class)
    
    Returns:
        heatmap: Grad-CAM heatmap
    """
    # Create a model that maps the input image to the activations
    # of the last conv layer as well as the output predictions
    grad_model = tf.keras.models.Model(
        inputs=[model.inputs],
        outputs=[model.get_layer(last_conv_layer_name).output, model.output]
    )
    
    # Then compute the gradient of the top predicted class for our input image
    # with respect to the activations of the last conv layer
    with tf.GradientTape() as tape:
        last_conv_layer_output, preds = grad_model(img_array)
        if pred_index is None:
            pred_index = tf.argmax(preds[0])
        class_channel = preds[:, pred_index]
    
    # This is the gradient of the output neuron (top predicted or chosen)
    # with regard to the output feature map of the last conv layer
    grads = tape.gradient(class_channel, last_conv_layer_output)
    
    # This is a vector where each entry is the mean intensity of the gradient
    # over a specific feature map channel
    pooled_grads = tf.reduce_mean(grads, axis=(0, 1, 2))
    
    # We multiply each channel in the feature map array
    # by "how important this channel is" with regard to the top predicted class
    # then sum all the channels to obtain the heatmap class activation
    last_conv_layer_output = last_conv_layer_output[0]
    heatmap = last_conv_layer_output @ pooled_grads[..., tf.newaxis]
    heatmap = tf.squeeze(heatmap)
    
    # For visualization purpose, we will also normalize the heatmap between 0 & 1
    heatmap = tf.maximum(heatmap, 0) / tf.math.reduce_max(heatmap)
    
    return heatmap.numpy(), pred_index.numpy(), preds[0].numpy()

def overlay_heatmap_on_image(heatmap, original_img, alpha=0.4):
    """
    Overlay heatmap on original image
    
    Args:
        heatmap: Grad-CAM heatmap
        original_img: Original image (0-255)
        alpha: Transparency of heatmap overlay
    
    Returns:
        overlay: Image with heatmap overlay
    """
    # Resize heatmap to match original image
    heatmap = cv2.resize(heatmap, (original_img.shape[1], original_img.shape[0]))
    
    # Convert heatmap to RGB
    heatmap = np.uint8(255 * heatmap)
    heatmap = cv2.applyColorMap(heatmap, cv2.COLORMAP_JET)
    
    # Convert original image to BGR if needed
    if len(original_img.shape) == 2:
        original_img = cv2.cvtColor(original_img, cv2.COLOR_GRAY2BGR)
    elif original_img.shape[2] == 3:
        original_img = cv2.cvtColor(original_img.astype(np.uint8), cv2.COLOR_RGB2BGR)
    
    # Overlay heatmap
    superimposed_img = heatmap * alpha + original_img
    superimposed_img = np.clip(superimposed_img, 0, 255).astype(np.uint8)
    
    return superimposed_img

# =========================
# Evaluation Function (SAME as your original)
# =========================
def evaluate_model(model, test_data, class_names):
    """Evaluate model performance"""
    
    print("\n" + "="*60)
    print("EVALUATING MODEL")
    print("="*60)
    
    # Make predictions
    y_pred_proba = model.predict(test_data, verbose=0)
    y_pred = np.argmax(y_pred_proba, axis=1)
    
    # Get true labels
    y_true = []
    for _, batch_labels in test_data:
        y_true.extend(np.argmax(batch_labels.numpy(), axis=1))
    y_true = np.array(y_true)
    
    # Calculate metrics
    from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score
    
    accuracy = accuracy_score(y_true, y_pred)
    precision = precision_score(y_true, y_pred, average='weighted', zero_division=0)
    recall = recall_score(y_true, y_pred, average='weighted', zero_division=0)
    f1 = f1_score(y_true, y_pred, average='weighted', zero_division=0)
    
    print(f"\nPerformance Metrics:")
    print(f"Accuracy:  {accuracy:.4f}")
    print(f"Precision: {precision:.4f}")
    print(f"Recall:    {recall:.4f}")
    print(f"F1-Score:  {f1:.4f}")
    
    # Class-wise accuracy
    print(f"\nClass-wise Performance:")
    print(f"{'Class Name':<20} {'Accuracy':<10} {'Samples':<10}")
    print(f"{'-'*50}")
    
    for i in range(len(class_names)):
        class_mask = (y_true == i)
        if np.sum(class_mask) > 0:
            class_acc = accuracy_score(y_true[class_mask], y_pred[class_mask])
            samples = np.sum(class_mask)
            print(f"{class_names[i]:<20} {class_acc:<10.4f} {samples:<10}")
    
    # Confusion matrix
    cm = confusion_matrix(y_true, y_pred)
    
    plt.figure(figsize=(12, 10))
    sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', 
                xticklabels=class_names, yticklabels=class_names)
    plt.title('Confusion Matrix', fontsize=16, fontweight='bold')
    plt.xlabel('Predicted Label', fontweight='bold')
    plt.ylabel('True Label', fontweight='bold')
    plt.xticks(rotation=45, ha='right')
    plt.yticks(rotation=0)
    plt.tight_layout()
    plt.show()
    
    return y_true, y_pred, y_pred_proba

# Evaluate the model
y_true, y_pred, y_pred_proba = evaluate_model(model, test_dataset, class_names)

# =========================
# Grad-CAM Visualizations
# =========================
def visualize_gradcam_for_samples(X_samples, y_true_samples, y_pred_samples, 
                                 class_names, model, num_samples=6):
    """
    Visualize Grad-CAM for samples
    
    Args:
        X_samples: Sample images
        y_true_samples: True labels for samples
        y_pred_samples: Predicted labels for samples
        class_names: List of class names
        model: Trained model
        num_samples: Number of samples to visualize
    """
    
    print("\n" + "="*60)
    print("GRAD-CAM VISUALIZATIONS")
    print("="*60)
    
    # Create figure
    fig, axes = plt.subplots(num_samples, 4, figsize=(16, 4 * num_samples))
    
    if num_samples == 1:
        axes = axes.reshape(1, -1)
    
    for i in range(num_samples):
        idx = i  # Use first n samples
        
        # Get image and labels
        img_array = X_samples[idx]
        original_img = (img_array * 255).astype(np.uint8)  # Convert back to 0-255
        true_class = y_true_samples[idx]
        pred_class = y_pred_samples[idx]
        
        # Prepare image for model
        img_for_model = np.expand_dims(img_array, axis=0)
        
        # Get heatmap using Grad-CAM
        # Note: 'densenet121' is the name of the base model layer
        heatmap, _, pred_probs = make_gradcam_heatmap(
            img_for_model, 
            model, 
            'densenet121',  # Last conv layer name
            pred_class
        )
        
        # Create overlay
        overlay = overlay_heatmap_on_image(heatmap, original_img)
        
        # Convert for display
        overlay_rgb = cv2.cvtColor(overlay, cv2.COLOR_BGR2RGB)
        original_img_rgb = cv2.cvtColor(original_img, cv2.COLOR_BGR2RGB)
        
        # Plot original image
        axes[i, 0].imshow(original_img_rgb)
        axes[i, 0].set_title(f'True: {class_names[true_class]}', fontsize=10, fontweight='bold')
        axes[i, 0].axis('off')
        
        # Plot heatmap
        axes[i, 1].imshow(heatmap, cmap='jet')
        axes[i, 1].set_title('Grad-CAM Heatmap', fontsize=10, fontweight='bold')
        axes[i, 1].axis('off')
        
        # Plot overlay
        axes[i, 2].imshow(overlay_rgb)
        axes[i, 2].set_title(f'Pred: {class_names[pred_class]}', fontsize=10, fontweight='bold')
        axes[i, 2].axis('off')
        
        # Plot top predictions
        top_k = min(5, len(class_names))
        top_indices = np.argsort(pred_probs)[-top_k:][::-1]
        top_probs = pred_probs[top_indices]
        top_classes = [class_names[idx] for idx in top_indices]
        
        colors = plt.cm.RdYlGn(top_probs / top_probs.max())
        axes[i, 3].barh(range(top_k), top_probs, color=colors)
        axes[i, 3].set_yticks(range(top_k))
        axes[i, 3].set_yticklabels(top_classes, fontsize=9)
        axes[i, 3].set_xlabel('Probability', fontsize=9)
        axes[i, 3].set_title('Top Predictions', fontsize=10, fontweight='bold')
        axes[i, 3].invert_yaxis()
        
        # Add correctness indicator
        if pred_class == true_class:
            axes[i, 0].text(0.5, -0.1, '✓ CORRECT', transform=axes[i, 0].transAxes,
                           ha='center', fontsize=10, fontweight='bold', color='green')
        else:
            axes[i, 0].text(0.5, -0.1, '✗ WRONG', transform=axes[i, 0].transAxes,
                           ha='center', fontsize=10, fontweight='bold', color='red')
    
    plt.suptitle('Grad-CAM Visualizations - Understanding Model Decisions', 
                fontsize=16, fontweight='bold', y=1.02)
    plt.tight_layout()
    plt.show()

# Get some samples from test set
test_samples = 6
X_samples = X_test[:test_samples]
y_true_samples = np.argmax(y_test_cat[:test_samples], axis=1)
y_pred_samples = y_pred[:test_samples]

# Visualize Grad-CAM
visualize_gradcam_for_samples(X_samples, y_true_samples, y_pred_samples, 
                             class_names, model, num_samples=test_samples)

# =========================
# Analyze Specific Cases
# =========================
def analyze_correct_vs_incorrect(X_test, y_true, y_pred, class_names, model):
    """
    Analyze correct and incorrect predictions with Grad-CAM
    """
    
    # Find correct and incorrect predictions
    correct_indices = np.where(y_true == y_pred)[0]
    incorrect_indices = np.where(y_true != y_pred)[0]
    
    print(f"\nPrediction Analysis:")
    print(f"Correct predictions: {len(correct_indices)} / {len(y_true)} ({len(correct_indices)/len(y_true)*100:.1f}%)")
    print(f"Incorrect predictions: {len(incorrect_indices)} / {len(y_true)} ({len(incorrect_indices)/len(y_true)*100:.1f}%)")
    
    # Analyze a correct prediction
    if len(correct_indices) > 0:
        print(f"\nAnalyzing a CORRECT prediction...")
        idx = correct_indices[0]
        
        img_array = X_test[idx]
        original_img = (img_array * 255).astype(np.uint8)
        true_class = y_true[idx]
        pred_class = y_pred[idx]
        
        # Prepare image for model
        img_for_model = np.expand_dims(img_array, axis=0)
        
        # Get heatmap
        heatmap, _, pred_probs = make_gradcam_heatmap(
            img_for_model, 
            model, 
            'densenet121',
            pred_class
        )
        
        # Create overlay
        overlay = overlay_heatmap_on_image(heatmap, original_img)
        overlay_rgb = cv2.cvtColor(overlay, cv2.COLOR_BGR2RGB)
        original_img_rgb = cv2.cvtColor(original_img, cv2.COLOR_BGR2RGB)
        
        # Create visualization
        fig, axes = plt.subplots(1, 4, figsize=(16, 4))
        
        # Original image
        axes[0].imshow(original_img_rgb)
        axes[0].set_title(f'True: {class_names[true_class]}', fontsize=12, fontweight='bold')
        axes[0].axis('off')
        
        # Heatmap
        axes[1].imshow(heatmap, cmap='jet')
        axes[1].set_title('Attention Heatmap', fontsize=12, fontweight='bold')
        axes[1].axis('off')
        
        # Overlay
        axes[2].imshow(overlay_rgb)
        axes[2].set_title(f'Pred: {class_names[pred_class]}', fontsize=12, fontweight='bold')
        axes[2].axis('off')
        
        # Top predictions
        top_k = 3
        top_indices = np.argsort(pred_probs)[-top_k:][::-1]
        top_probs = pred_probs[top_indices]
        top_classes = [class_names[idx] for idx in top_indices]
        
        colors = plt.cm.RdYlGn(top_probs / top_probs.max())
        axes[3].barh(range(top_k), top_probs, color=colors)
        axes[3].set_yticks(range(top_k))
        axes[3].set_yticklabels(top_classes, fontsize=10)
        axes[3].set_xlabel('Probability', fontsize=10)
        axes[3].set_title('Top Predictions', fontsize=12, fontweight='bold')
        axes[3].invert_yaxis()
        
        plt.suptitle(f'CORRECT Prediction: Model focuses on discriminative features', 
                    fontsize=14, fontweight='bold', y=1.05)
        plt.tight_layout()
        plt.show()
    
    # Analyze an incorrect prediction
    if len(incorrect_indices) > 0:
        print(f"\nAnalyzing an INCORRECT prediction...")
        idx = incorrect_indices[0]
        
        img_array = X_test[idx]
        original_img = (img_array * 255).astype(np.uint8)
        true_class = y_true[idx]
        pred_class = y_pred[idx]
        
        # Prepare image for model
        img_for_model = np.expand_dims(img_array, axis=0)
        
        # Get heatmap for predicted class
        heatmap_pred, _, pred_probs = make_gradcam_heatmap(
            img_for_model, 
            model, 
            'densenet121',
            pred_class
        )
        
        # Create overlay
        overlay = overlay_heatmap_on_image(heatmap_pred, original_img)
        overlay_rgb = cv2.cvtColor(overlay, cv2.COLOR_BGR2RGB)
        original_img_rgb = cv2.cvtColor(original_img, cv2.COLOR_BGR2RGB)
        
        # Create visualization
        fig, axes = plt.subplots(1, 4, figsize=(16, 4))
        
        # Original image
        axes[0].imshow(original_img_rgb)
        axes[0].set_title(f'True: {class_names[true_class]}', fontsize=12, fontweight='bold')
        axes[0].axis('off')
        
        # Heatmap for predicted class
        axes[1].imshow(heatmap_pred, cmap='jet')
        axes[1].set_title(f'Heatmap for {class_names[pred_class]}', fontsize=12, fontweight='bold')
        axes[1].axis('off')
        
        # Overlay
        axes[2].imshow(overlay_rgb)
        axes[2].set_title(f'Predicted as: {class_names[pred_class]}', fontsize=12, fontweight='bold')
        axes[2].axis('off')
        
        # Confidence scores
        classes_to_show = [true_class, pred_class]
        confidences = [pred_probs[true_class], pred_probs[pred_class]]
        class_names_to_show = [class_names[true_class], class_names[pred_class]]
        
        colors = ['red' if i == pred_class else 'blue' for i in [true_class, pred_class]]
        axes[3].barh(range(2), confidences, color=colors)
        axes[3].set_yticks(range(2))
        axes[3].set_yticklabels(class_names_to_show, fontsize=10)
        axes[3].set_xlabel('Probability', fontsize=10)
        axes[3].set_title('Model Confidence', fontsize=12, fontweight='bold')
        axes[3].invert_yaxis()
        
        plt.suptitle(f'INCORRECT Prediction: Model confuses {class_names[true_class]} with {class_names[pred_class]}', 
                    fontsize=14, fontweight='bold', y=1.05)
        plt.tight_layout()
        plt.show()
        
        print(f"True class ({class_names[true_class]}) probability: {pred_probs[true_class]:.2%}")
        print(f"Predicted class ({class_names[pred_class]}) probability: {pred_probs[pred_class]:.2%}")

# Analyze specific cases
analyze_correct_vs_incorrect(X_test, y_true, y_pred, class_names, model)

# =========================
# Save Model and Results
# =========================
print("\n" + "="*60)
print("SAVING MODEL AND RESULTS")
print("="*60)

# Save the model
model.save('densenet121_with_gradcam_final.h5')
print("Model saved as 'densenet121_with_gradcam_final.h5'")

# Save class names
np.save('class_names.npy', class_names)
print("Class names saved as 'class_names.npy'")

# Save predictions
from sklearn.metrics import accuracy_score
results = {
    'y_true': y_true,
    'y_pred': y_pred,
    'y_pred_proba': y_pred_proba,
    'class_names': class_names,
    'accuracy': accuracy_score(y_true, y_pred),
    'training_time': training_time
}
np.save('gradcam_analysis_results.npy', results)
print("Results saved as 'gradcam_analysis_results.npy'")

# =========================
# Training History Visualization
# =========================
print("\nGenerating training history visualization...")

plt.figure(figsize=(12, 4))

# Plot accuracy
plt.subplot(1, 2, 1)
plt.plot(history.history['accuracy'], label='Train Accuracy', linewidth=2)
plt.plot(history.history['val_accuracy'], label='Validation Accuracy', linewidth=2)
plt.title('Model Accuracy', fontsize=14, fontweight='bold')
plt.xlabel('Epoch', fontweight='bold')
plt.ylabel('Accuracy', fontweight='bold')
plt.legend()
plt.grid(True, alpha=0.3)

# Plot loss
plt.subplot(1, 2, 2)
plt.plot(history.history['loss'], label='Train Loss', linewidth=2)
plt.plot(history.history['val_loss'], label='Validation Loss', linewidth=2)
plt.title('Model Loss', fontsize=14, fontweight='bold')
plt.xlabel('Epoch', fontweight='bold')
plt.ylabel('Loss', fontweight='bold')
plt.legend()
plt.grid(True, alpha=0.3)

plt.suptitle('DenseNet121 Training History (with Grad-CAM)', fontsize=16, fontweight='bold', y=1.05)
plt.tight_layout()
plt.show()

print("\n" + "="*60)
print("GRAD-CAM IMPLEMENTATION COMPLETE!")
print("="*60)
print("\nSummary:")
print(f"1. Model trained with same parameters as original DenseNet121")
print(f"2. Test Accuracy: {accuracy_score(y_true, y_pred):.4f}")
print(f"3. Training Time: {training_time:.2f} seconds")
print(f"4. Grad-CAM visualizations generated")
print(f"5. Analysis of correct/incorrect predictions completed")
print("="*60)
print("\nKey Features:")
print("✓ Same architecture and training parameters as original")
print("✓ Functional API for proper Grad-CAM access")
print("✓ Heatmap generation for model interpretability")
print("✓ Correct vs incorrect prediction analysis")
print("✓ All results saved for future reference")
print("="*60)
```
